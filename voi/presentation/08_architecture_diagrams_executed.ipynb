# Model Architecture Diagrams for Presentation

Large, clear visualizations of model architectures optimized for presentation display

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrow
import numpy as np
import warnings
warnings.filterwarnings('ignore')

plt.rcParams['figure.dpi'] = 150
plt.rcParams['savefig.dpi'] = 300
```

## V4 Ensemble Architecture (Fixed Layout with Larger Text)

```python
fig, ax = plt.subplots(figsize=(16, 10))

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Input features (left side)
input_features = [
    "PM2.5 Lags (1-168h)",
    "Rolling Stats (3-48h)",
    "Weather Features",
    "Traffic Features",
    "Temporal Encoding",
    "Spatial (Hexagon ID)"
]

for i, feature in enumerate(input_features):
    y_pos = 8.5 - i * 1.2
    rect = FancyBboxPatch((0.5, y_pos), 1.8, 0.8, 
                          boxstyle="round,pad=0.05",
                          facecolor='#E8F4FD', edgecolor='#2E86AB', linewidth=2.5)
    ax.add_patch(rect)
    ax.text(1.4, y_pos + 0.4, feature, ha='center', va='center', 
           fontsize=13, fontweight='bold')

# Three models (center)
models = [
    ("LightGBM", "#4CAF50", "Tree-based\nGradient Boosting"),
    ("XGBoost", "#2196F3", "Extreme\nGradient Boosting"),
    ("CatBoost", "#FF9800", "Categorical\nBoosting")
]

for i, (name, color, desc) in enumerate(models):
    x_pos = 4 + i * 2
    rect = FancyBboxPatch((x_pos - 0.7, 3), 1.4, 2, 
                          boxstyle="round,pad=0.05",
                          facecolor=color, alpha=0.3, edgecolor=color, linewidth=2.5)
    ax.add_patch(rect)
    ax.text(x_pos, 4.5, name, ha='center', va='center', 
           fontsize=15, fontweight='bold')
    ax.text(x_pos, 3.5, desc, ha='center', va='center', 
           fontsize=11, style='italic')

# Draw arrows from features to models
for i in range(6):
    y_start = 8.5 - i * 1.2 + 0.4
    for j in range(3):
        x_end = 4 + j * 2 - 0.7
        ax.arrow(2.3, y_start, x_end - 2.3 - 0.1, 4 - y_start, 
                head_width=0.12, head_length=0.06, fc='gray', ec='gray', 
                alpha=0.3, linewidth=1.5)

# Weighted Ensemble box (bottom) - EXTENDED TO SPAN ALL MODELS
ensemble_rect = FancyBboxPatch((3, 0.5), 4, 1.2, 
                               boxstyle="round,pad=0.05",
                               facecolor='#F44336', alpha=0.3, 
                               edgecolor='#F44336', linewidth=3)
ax.add_patch(ensemble_rect)
ax.text(5, 1.1, "Weighted\nEnsemble", ha='center', va='center', 
       fontsize=15, fontweight='bold')

# Draw arrows from models to ensemble
for i in range(3):
    x_start = 4 + i * 2
    ax.arrow(x_start, 3, 0, -1.2, head_width=0.18, head_length=0.12, 
            fc='gray', ec='gray', alpha=0.5, linewidth=2)

# Title and subtitle
ax.text(5, 9.5, "Ensemble Model Architecture", ha='center', va='center', 
       fontsize=20, fontweight='bold')
ax.text(5, 9, "69 Features → 3 Models → Weighted Average", ha='center', va='center', 
       fontsize=14, style='italic')

plt.tight_layout()
plt.savefig('graphs/architecture/ensemble_architecture_presentation.png', 
           dpi=300, bbox_inches='tight', facecolor='white')
print("Saved: graphs/architecture/ensemble_architecture_presentation.png")
plt.show()
```

## V1 Linear Regression Architecture (Simple Baseline)

```python
fig, ax = plt.subplots(figsize=(16, 10))

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Single input feature box (PM2.5 lags only)
y_pos = 5
rect = FancyBboxPatch((0.5, y_pos - 0.6), 2.2, 1.2, 
                      boxstyle="round,pad=0.05",
                      facecolor='#E8F4FD', edgecolor='#2E86AB', linewidth=3)
ax.add_patch(rect)
ax.text(1.6, y_pos, "PM2.5 Lags\n(1-168h)", ha='center', va='center', 
       fontsize=14, fontweight='bold')

# Linear Regression model (center)
model_rect = FancyBboxPatch((3.8, 3.8), 2.4, 2.4, 
                           boxstyle="round,pad=0.05",
                           facecolor='#90CAF9', alpha=0.4, 
                           edgecolor='#1565C0', linewidth=3)
ax.add_patch(model_rect)
ax.text(5, 5.5, "Linear\nRegression", ha='center', va='center', 
       fontsize=16, fontweight='bold')
ax.text(5, 4.3, "Simple baseline\nmodel", ha='center', va='center', 
       fontsize=12, style='italic')

# Output prediction (right)
output_rect = FancyBboxPatch((7.3, y_pos - 0.6), 2.2, 1.2, 
                            boxstyle="round,pad=0.05",
                            facecolor='#FFE0B2', edgecolor='#F57C00', linewidth=3)
ax.add_patch(output_rect)
ax.text(8.4, y_pos, "PM2.5\nPrediction", ha='center', va='center', 
       fontsize=14, fontweight='bold')

# Arrow from input to model
ax.arrow(2.7, y_pos, 1.0, 0, head_width=0.2, head_length=0.15, 
        fc='#666', ec='#666', linewidth=2.5)

# Arrow from model to output
ax.arrow(6.2, y_pos, 1.0, 0, head_width=0.2, head_length=0.15, 
        fc='#666', ec='#666', linewidth=2.5)

# Title and subtitle
ax.text(5, 8.5, "V1 Baseline Model Architecture", ha='center', va='center', 
       fontsize=20, fontweight='bold')
ax.text(5, 7.9, "PM2.5 Time Series Only → Linear Regression → Direct Prediction", 
       ha='center', va='center', fontsize=14, style='italic')

# Performance annotation
ax.text(5, 2, "MAE: 2.35-5.85 μg/m³ (1-24h)\nR²: 0.77 to -0.06 (1-24h)", 
       ha='center', va='center', fontsize=12, 
       bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))

plt.tight_layout()
plt.savefig('graphs/architecture/v1_linear_regression_architecture_presentation.png', 
           dpi=300, bbox_inches='tight', facecolor='white')
print("Saved: graphs/architecture/v1_linear_regression_architecture_presentation.png")
plt.show()
```

## V4 Baseline Architecture (Linear Regression with Features)

```python
fig, ax = plt.subplots(figsize=(16, 10))

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Input features (left side) - simplified for V4 baseline
input_features = [
    "PM2.5 Lags (1-168h)",
    "Weather Features",
    "Traffic Features",
    "Temporal Encoding"
]

for i, feature in enumerate(input_features):
    y_pos = 7.5 - i * 1.5
    rect = FancyBboxPatch((0.5, y_pos), 2.0, 1.0, 
                          boxstyle="round,pad=0.05",
                          facecolor='#E8F4FD', edgecolor='#2E86AB', linewidth=2.5)
    ax.add_patch(rect)
    ax.text(1.5, y_pos + 0.5, feature, ha='center', va='center', 
           fontsize=13, fontweight='bold')

# Linear Regression model (center)
model_rect = FancyBboxPatch((3.8, 3.8), 2.4, 2.4, 
                           boxstyle="round,pad=0.05",
                           facecolor='#A5D6A7', alpha=0.4, 
                           edgecolor='#2E7D32', linewidth=3)
ax.add_patch(model_rect)
ax.text(5, 5.5, "Linear\nRegression", ha='center', va='center', 
       fontsize=16, fontweight='bold')
ax.text(5, 4.3, "Enhanced with\nweather & traffic", ha='center', va='center', 
       fontsize=12, style='italic')

# Output prediction (right)
output_rect = FancyBboxPatch((7.3, 4.4), 2.2, 1.2, 
                            boxstyle="round,pad=0.05",
                            facecolor='#FFE0B2', edgecolor='#F57C00', linewidth=3)
ax.add_patch(output_rect)
ax.text(8.4, 5, "PM2.5\nPrediction", ha='center', va='center', 
       fontsize=14, fontweight='bold')

# Arrows from features to model
for i in range(4):
    y_start = 7.5 - i * 1.5 + 0.5
    ax.arrow(2.5, y_start, 1.2, 5 - y_start, 
            head_width=0.15, head_length=0.1, 
            fc='gray', ec='gray', alpha=0.4, linewidth=1.5)

# Arrow from model to output
ax.arrow(6.2, 5, 1.0, 0, head_width=0.2, head_length=0.15, 
        fc='#666', ec='#666', linewidth=2.5)

# Title and subtitle
ax.text(5, 8.5, "V4 Baseline Model Architecture", ha='center', va='center', 
       fontsize=20, fontweight='bold')
ax.text(5, 7.9, "Multiple Features → Linear Regression → Improved Prediction", 
       ha='center', va='center', fontsize=14, style='italic')

# Performance annotation
ax.text(5, 2, "MAE: 2.08-4.84 μg/m³ (1-24h)\nR²: 0.85 to 0.20 (1-24h)\n~20% improvement over V1", 
       ha='center', va='center', fontsize=12, 
       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))

plt.tight_layout()
plt.savefig('graphs/architecture/v4_baseline_architecture_presentation.png', 
           dpi=300, bbox_inches='tight', facecolor='white')
print("Saved: graphs/architecture/v4_baseline_architecture_presentation.png")
plt.show()
```

## Side-by-Side Model Evolution Comparison

```python
fig, axes = plt.subplots(1, 3, figsize=(20, 8))

# V1 Baseline (Left)
ax = axes[0]
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Input
rect = FancyBboxPatch((1, 4.5), 2, 1, boxstyle="round,pad=0.05",
                      facecolor='#E8F4FD', edgecolor='#2E86AB', linewidth=2)
ax.add_patch(rect)
ax.text(2, 5, "PM2.5 Lags", ha='center', va='center', fontsize=11, fontweight='bold')

# Model
rect = FancyBboxPatch((4.5, 4), 2, 2, boxstyle="round,pad=0.05",
                      facecolor='#90CAF9', alpha=0.4, edgecolor='#1565C0', linewidth=2)
ax.add_patch(rect)
ax.text(5.5, 5, "Linear\nRegression", ha='center', va='center', fontsize=12, fontweight='bold')

# Arrow
ax.arrow(3, 5, 1.4, 0, head_width=0.2, head_length=0.1, fc='gray', ec='gray')

ax.text(5, 8, "V1 Baseline", ha='center', fontsize=14, fontweight='bold')
ax.text(5, 2, "MAE: 2.35 μg/m³ (1h)\nR²: 0.769", ha='center', fontsize=10,
       bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.5))

# V4 Baseline (Center)
ax = axes[1]
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Multiple inputs
features = ["PM2.5", "Weather", "Traffic"]
for i, feat in enumerate(features):
    y = 6 - i * 1.5
    rect = FancyBboxPatch((0.5, y - 0.4), 1.8, 0.8, boxstyle="round,pad=0.05",
                          facecolor='#E8F4FD', edgecolor='#2E86AB', linewidth=2)
    ax.add_patch(rect)
    ax.text(1.4, y, feat, ha='center', va='center', fontsize=10)
    ax.arrow(2.3, y, 1.9, 5 - y, head_width=0.1, head_length=0.05, 
            fc='gray', ec='gray', alpha=0.3)

# Model
rect = FancyBboxPatch((4.5, 4), 2, 2, boxstyle="round,pad=0.05",
                      facecolor='#A5D6A7', alpha=0.4, edgecolor='#2E7D32', linewidth=2)
ax.add_patch(rect)
ax.text(5.5, 5, "Linear\nRegression", ha='center', va='center', fontsize=12, fontweight='bold')

ax.text(5, 8, "V4 Baseline", ha='center', fontsize=14, fontweight='bold')
ax.text(5, 2, "MAE: 2.08 μg/m³ (1h)\nR²: 0.845\n↓11% MAE", ha='center', fontsize=10,
       bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.5))

# V4 Ensemble (Right)
ax = axes[2]
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Features to models
features = ["69 Features"]
rect = FancyBboxPatch((0.5, 4.5), 1.8, 1, boxstyle="round,pad=0.05",
                      facecolor='#E8F4FD', edgecolor='#2E86AB', linewidth=2)
ax.add_patch(rect)
ax.text(1.4, 5, "69 Features", ha='center', va='center', fontsize=10, fontweight='bold')

# Three models
models = [("LGB", "#4CAF50"), ("XGB", "#2196F3"), ("Cat", "#FF9800")]
for i, (name, color) in enumerate(models):
    x = 3.5 + i * 1.5
    rect = FancyBboxPatch((x - 0.4, 4.5), 0.8, 1, boxstyle="round,pad=0.05",
                          facecolor=color, alpha=0.3, edgecolor=color, linewidth=2)
    ax.add_patch(rect)
    ax.text(x, 5, name, ha='center', va='center', fontsize=9, fontweight='bold')
    ax.arrow(2.3, 5, x - 2.3 - 0.4, 0, head_width=0.1, head_length=0.05,
            fc='gray', ec='gray', alpha=0.3)
    ax.arrow(x, 4.5, 0, -1.3, head_width=0.1, head_length=0.05,
            fc='gray', ec='gray', alpha=0.3)

# Ensemble
rect = FancyBboxPatch((3, 2.5), 3.5, 0.7, boxstyle="round,pad=0.05",
                      facecolor='#F44336', alpha=0.3, edgecolor='#F44336', linewidth=2)
ax.add_patch(rect)
ax.text(4.75, 2.85, "Ensemble", ha='center', va='center', fontsize=10, fontweight='bold')

ax.text(5, 8, "V4 Ensemble", ha='center', fontsize=14, fontweight='bold')
ax.text(5, 1.5, "MAE: 2.00 μg/m³ (1h)\nR²: 0.824\n↓15% MAE", ha='center', fontsize=10,
       bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.5))

plt.suptitle("Model Evolution: Architecture Comparison", fontsize=18, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('graphs/architecture/model_evolution_comparison.png', 
           dpi=300, bbox_inches='tight', facecolor='white')
print("Saved: graphs/architecture/model_evolution_comparison.png")
plt.show()
```
